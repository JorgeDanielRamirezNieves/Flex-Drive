### **📌Limpieza de la Pregunta 7: Actividades Actuales**

Esta pregunta es de selección múltiple, por lo tanto, cada respuesta puede contener varias opciones.
Pasos a seguir:
1️⃣ Definir las categorías oficiales según el formulario original.
2️⃣ Limpiar cada respuesta y convertirla en una lista organizada.
3️⃣ Unificar términos similares y eliminar inconsistencias.

Esto nos permitirá analizar de forma precisa las ocupaciones actuales de los encuestados.

```python
import pandas as pd
from collections import Counter
import matplotlib.pyplot as plt

# --- Cargar archivo ---
df = pd.read_excel('../../docs/FLEX DRIVE - Permítenos Conocerte.xlsx', sheet_name='Sheet1')

# --- LIMPIEZA DE P7: Actividades Actuales ---
def limpiar_actividad_v2(respuesta):
    if pd.isna(respuesta):
        return []

    actividades = [a.strip().rstrip('.').lower() for a in respuesta.split(';') if a.strip() != '']
    actividades_limpias = []

    for act in actividades:
        if 'secundaria' in act:
            actividades_limpias.append('Estudiante secundaria')
        elif 'universitario' in act or 'sena' in act or 'aprendiz' in act:
            actividades_limpias.append('Estudiante universitario o de estudios superiores')
        elif 'trabajador dependiente' in act or 'empleada' in act or 'empleado publico' in act or 'empleado y estudiante' in act or 'empleado comerciante' in act or 'concejal' in act:
            actividades_limpias.append('Trabajador dependiente')
        elif 'independiente' in act or 'freelancer' in act or 'consultor' in act:
            actividades_limpias.append('Independiente o freelancer')
        elif 'empresario' in act:
            actividades_limpias.append('Empresario')
        elif 'desempleado' in act:
            actividades_limpias.append('Desempleado')
        elif 'descanso' in act or 'sabático' in act:
            actividades_limpias.append('Descanso / sabático')
        elif 'jubilado' in act or 'pensionado' in act or 'retirado' in act:
            actividades_limpias.append('Jubilado / pensionado / retirado')
        elif 'hogar' in act or 'ama de casa' in act or 'oficios varios' in act or 'cuido' in act or 'labores domésticas' in act:
            actividades_limpias.append('Ama de casa')
        # No añadimos categoría "Otro"

    return list(set(actividades_limpias))

# Aplicar limpieza
df['P7-Actividad_Limpia'] = df['P7-Actividad_Actual_Multi'].apply(limpiar_actividad_v2)

# --- Conteo de Categorías ---
conteo = Counter([actividad for sublista in df['P7-Actividad_Limpia'] for actividad in sublista])

print("\n🔢 Conteo total de cada categoría:")
for categoria, total in conteo.items():
    print(f"{categoria}: {total}")

# --- Visualización ---
labels, values = zip(*conteo.items())

plt.figure(figsize=(10, 6))
plt.bar(labels, values, color='mediumturquoise')
plt.xticks(rotation=45, ha='right')
plt.title("📊 Actividades Actuales de los Encuestados")
plt.ylabel("Frecuencia")
plt.tight_layout()
plt.show()
```

### **📌 Limpieza de la Pregunta 10: Servicios de Transporte Público Utilizados**

Esta pregunta es de **selección múltiple**, por lo tanto, cada encuestado puede haber elegido varios medios de transporte a la vez.

**Pasos a seguir:**  
- 1️⃣ Definir las *categorías oficiales* según las opciones del formulario.  
- 2️⃣ *Limpiar y separar* adecuadamente las respuestas múltiples.  
- 3️⃣ Agrupar y normalizar términos similares, incluyendo aquellos que caen en la categoría "**Otras**".  

🔍 *El objetivo es identificar con claridad cuáles son los medios de transporte más utilizados actualmente por los encuestados, lo cual es fundamental para comprender sus hábitos de movilidad y diseñar servicios acordes a sus necesidades.*

```python
# 📦 Paso 1: Función final para limpiar y clasificar las respuestas múltiples
def limpiar_transporte_publico_final(respuesta):
    if pd.isna(respuesta):
        return []

    items = [r.strip().lower().rstrip('.') for r in respuesta.split(';') if r.strip()]
    categorias = []

    for item in items:
        if 'taxis' in item or item == 'taxi':
            categorias.append('Taxis')
        elif 'buses municipales' in item or item == 'bus':
            categorias.append('Buses municipales')
        elif 'buses intermunicipales' in item or 'piratas' in item or 'coontrasven' in item:
            categorias.append('Buses intermunicipales')
        elif 'metro' in item:
            categorias.append('Metro')
        elif 'tele-cable' in item:
            categorias.append('Tele-cable')
        elif 'sitp' in item or 'interconexión' in item or 'transmilenio' in item or 'metrolínea' in item or 'transcaribe' in item or 'mio' in item:
            categorias.append('SITP')
        elif 'uber' in item or 'didi' in item or 'indriver' in item or 'in driver' in item or 'plataforma' in item or 'aplicaciones' in item:
            categorias.append('Plataformas digitales')
        elif 'bicicleta' in item or 'cicla' in item or 'caminando' in item or 'moto' in item or 'mototaxi' in item:
            categorias.append('Alternativo / ecológico')
        elif 'carro' in item or 'auto propio' in item or 'drive' in item or item in ['na', 'ninguna en el momento']:
            continue  # Ignorar
        else:
            categorias.append('Otro')

    return list(set(categorias))  # Evitar duplicados

# 🧹 Paso 2: Aplicar la limpieza a la columna de la pregunta 10
df['P10-Transporte_Limpio'] = df['P10-Servicios_Públicos_Usados_Multi'].apply(limpiar_transporte_publico_final)

# 📊 Paso 3: Contar las respuestas por categoría
from collections import Counter

conteo = Counter([item for sublist in df['P10-Transporte_Limpio'] for item in sublist])
print("🔢 Conteo actualizado de servicios públicos utilizados:")
for categoria, total in conteo.items():
    print(f"{categoria}: {total}")

# 🔍 Paso 4: Revisar ejemplos de respuestas marcadas como "Otro"
otros_transporte = df[df['P10-Transporte_Limpio'].apply(lambda x: 'Otro' in x)]
display(otros_transporte[['P10-Servicios_Públicos_Usados_Multi', 'P10-Transporte_Limpio']].head(20))

# 📌 Paso 5: Extraer respuestas originales únicas que entraron en "Otro"
respuestas_otros = otros_transporte['P10-Servicios_Públicos_Usados_Multi'].unique()

# 📈 Paso 6: Visualizar los resultados con un gráfico de barras
import matplotlib.pyplot as plt

labels = list(conteo.keys())
values = list(conteo.values())

plt.figure(figsize=(10, 6))
plt.bar(labels, values, color='skyblue')
plt.title('Uso de Servicios de Transporte Público')
plt.xlabel('Tipo de transporte')
plt.ylabel('Número de personas')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

## 🚶‍♂️ Pregunta 11: ¿Cuál es el motivo por el que te desplazas con regularidad?

Esta pregunta permite conocer las razones principales por las cuales las personas se movilizan habitualmente. Se trata de una **pregunta de selección múltiple**, por lo que los encuestados pudieron marcar varias opciones al mismo tiempo.

### 📝 Opciones de respuesta disponibles:
- Por trabajo
- Por estudio
- Por compromisos personales o familiares
- Por turismo
- Otras

### 🎯 Objetivo del análisis
El propósito de este análisis es **identificar los patrones de desplazamiento** en función de las actividades cotidianas más frecuentes. Esto nos permite entender mejor las **necesidades de movilidad** de los usuarios, así como posibles oportunidades para ofrecer servicios diferenciados según el motivo del viaje.

Posteriormente, se aplicará un proceso de limpieza y categorización similar al de otras preguntas, con el fin de contar, clasificar y visualizar los distintos motivos mencionados, incluyendo aquellos registrados bajo la categoría **"Otras"**, que serán inspeccionados para una mejor comprensión de las respuestas abiertas.

```python
df_FlexDrive['P11-Motivo_Desplazamiento_Multi'].unique()
```

```python
df_FlexDrive['P11-Motivo_Desplazamiento_Multi'].value_counts()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
from collections import Counter

# Cargar archivo
df = pd.read_excel('../../docs/FLEX DRIVE - Permítenos Conocerte.xlsx')

# --- 1. FUNCIÓN DE LIMPIEZA Y CLASIFICACIÓN ---
def limpiar_motivo_desplazamiento(respuesta):
    if pd.isna(respuesta):
        return []
    
    respuesta = respuesta.lower().strip()
    motivos_raw = [x.strip().rstrip('.') for x in respuesta.split(';') if x.strip()]
    motivos_limpios = []

    for motivo in motivos_raw:
        if 'trabajo' in motivo:
            motivos_limpios.append('Trabajo')
        elif 'estudio' in motivo:
            motivos_limpios.append('Estudio')
        elif 'compromisos personales' in motivo or 'familiares' in motivo:
            motivos_limpios.append('Compromisos personales/familiares')
        elif 'turismo' in motivo:
            motivos_limpios.append('Turismo')
        elif 'salud' in motivo or 'citas medicas' in motivo:
            motivos_limpios.append('Salud')
        elif 'comida' in motivo or 'entrenamientos' in motivo:
            motivos_limpios.append('Ocio/Entrenamientos')
        elif 'pico y placa' in motivo or 'lloviendo' in motivo:
            motivos_limpios.append('Condiciones externas')
        elif 'zona' in motivo or 'recoger a esposa' in motivo:
            motivos_limpios.append('Compromisos personales/familiares')
        else:
            motivos_limpios.append('Otro')

    return list(set(motivos_limpios))  # Elimina duplicados

# --- 2. APLICAR LA LIMPIEZA ---
df['P11_Motivo_Limpio'] = df['P11-Motivo_Desplazamiento_Multi'].apply(limpiar_motivo_desplazamiento)

# --- 3. CONTEO CATEGORÍAS ---
conteo_motivos = Counter([item for sublist in df['P11_Motivo_Limpio'] for item in sublist])

# --- 4. IMPRIMIR RESULTADOS ---
print("📊 Conteo final de motivos de desplazamiento:")
for motivo, cantidad in conteo_motivos.items():
    print(f"{motivo}: {cantidad}")

# --- 5. GRÁFICO DE BARRAS ---
plt.figure(figsize=(10,6))
plt.bar(conteo_motivos.keys(), conteo_motivos.values(), color='cornflowerblue')
plt.title("Motivos de desplazamiento con regularidad")
plt.xlabel("Motivo")
plt.ylabel("Cantidad de respuestas")
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
from collections import Counter

# --- 1. Cargar archivo ---
df = pd.read_excel('../../docs/FLEX DRIVE - Permítenos Conocerte.xlsx', sheet_name='Sheet1')

# --- 2. Diccionario de limpieza ---
diccionario_vehiculos_multi = {
    'Automóvil (carros, camionetas, entre otros)': 'Automóvil',
    'Moticicleta': 'Motocicleta',
    'Motocicleta;Automóvil (carros, camionetas, entre otros)': 'Automóvil;Motocicleta',
    'Bicicleta eléctrica': 'Bicicleta eléctrica',
    'Scooter': 'Scooter',
    'Camión': 'Camión',
    'Tráiler': 'Tráiler',
    'Autobús': 'Autobús'
}

# --- 3. Función de limpieza ---
def limpiar_vehiculos_multi(respuesta):
    if pd.isna(respuesta):
        return []

    categorias = []
    partes = [x.strip().rstrip(';') for x in respuesta.split(';') if x.strip()]
    for p in partes:
        if p in diccionario_vehiculos_multi:
            categorias.append(diccionario_vehiculos_multi[p])
        elif 'automóvil' in p.lower():
            categorias.append('Automóvil')
        elif 'moticicleta' in p.lower() or 'motocicleta' in p.lower():
            categorias.append('Motocicleta')
        elif 'bicicleta eléctrica' in p.lower():
            categorias.append('Bicicleta eléctrica')
        elif 'scooter' in p.lower():
            categorias.append('Scooter')
        elif 'camión' in p.lower():
            categorias.append('Camión')
        elif 'tráiler' in p.lower():
            categorias.append('Tráiler')
        elif 'autobús' in p.lower():
            categorias.append('Autobús')
        else:
            categorias.append('Otro')

    return list(set(categorias))  # eliminar duplicados

# --- 4. Aplicar limpieza ---
df['P19-Tipo_Vehiculos_Limpio'] = df['P17-Tipo_Veh]P15-Multi]'].apply(limpiar_vehiculos_multi)

# --- 5. Conteo final ---
conteo_multi = Counter([item for sublist in df['P19-Tipo_Vehiculos_Limpio'] for item in sublist])

print("📊 Conteo de tipos de vehículos múltiples:")
for tipo, cantidad in conteo_multi.items():
    print(f"{tipo}: {cantidad}")

# --- 6. Gráfico de barras ---
plt.figure(figsize=(10,6))
plt.bar(conteo_multi.keys(), conteo_multi.values(), color='mediumorchid')
plt.title("Tipos de Vehículos Poseídos (Selección Múltiple)")
plt.xlabel("Tipo de Vehículo")
plt.ylabel("Cantidad")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

# --- 7. Verificar qué quedó en "Otro" ---
otros = df[df['P19-Tipo_Vehiculos_Limpio'].apply(lambda x: 'Otro' in x)]
print("\n🔍 Registros clasificados como 'Otro':")
print(otros[['P17-Tipo_Veh]P15-Multi]']].dropna().head(10))
```

```python
import pandas as pd
import matplotlib.pyplot as plt
from collections import Counter

# Cargar archivo
df = pd.read_excel('../../docs/FLEX DRIVE - Permítenos Conocerte.xlsx', sheet_name='Sheet1')

# Paso 1: Función para limpiar y clasificar los métodos de pago
def limpiar_pagos(p):
    if pd.isna(p):
        return []
    
    opciones = [op.strip().rstrip('.') for op in p.split(';') if op.strip()]
    
    resultado = []
    for opcion in opciones:
        if 'en efectivo, antes de entregar el vehículo' in opcion.lower():
            resultado.append('Efectivo antes')
        elif 'en efectivo, cuando el cliente devuelve' in opcion.lower():
            resultado.append('Efectivo después')
        elif 'por transacción bancaria inmediata, una vez aprobado' in opcion.lower():
            resultado.append('Transferencia al aprobar')
        elif 'por transacción bancaria inmediata, al finalizar' in opcion.lower():
            resultado.append('Transferencia al finalizar')
        elif 'un respaldo con un pagaré automatizado' in opcion.lower():
            resultado.append('Pagaré Flex Drive')
        elif 'pague o se devuelva' in opcion.lower():
            continue  # Ignorar frases abiertas sin clasificar
        elif 'no rentaría mi vehiculo' in opcion.lower():
            continue  # No se clasifica
        else:
            resultado.append('Otro')  # Cualquier otro caso
    return resultado

# Paso 2: Aplicar la limpieza
df['P37_Pagos_Limpio'] = df['P37_Recibir_Pagos_Propietario[P30]'].apply(limpiar_pagos)

# Paso 3: Contar frecuencias
conteo_pagos = Counter([item for sublist in df['P37_Pagos_Limpio'] for item in sublist])

# Paso 4: Mostrar resultados
print("🔢 Conteo por forma de recibir pagos:\n")
for metodo, total in conteo_pagos.items():
    print(f"{metodo}: {total}")

# Paso 5: Gráfico de barras
plt.figure(figsize=(10, 6))
plt.bar(conteo_pagos.keys(), conteo_pagos.values(), color='mediumseagreen')
plt.title('Formas preferidas de recibir pagos - Flex Drive')
plt.xlabel('Método de pago')
plt.ylabel('Cantidad de respuestas')
plt.xticks(rotation=45)
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Cargar archivo Excel
df = pd.read_excel('../../docs/FLEX DRIVE - Permítenos Conocerte.xlsx', sheet_name='Sheet1')

# Paso 1: Función de limpieza
def clasificar_factores_confianza(valor):
    if pd.isna(valor):
        return 'Sin respuesta'
    
    texto = valor.strip().lower()
    if texto in ['ninguna', 'ninguno', 'ninguna ', 'ningún otro', 'no se', 'no aplica', '…', '...', '.', 'sii']:
        return 'Sin respuesta'
    
    if any(palabra in texto for palabra in ['espejo', 'direccional', 'llantas', 'limpieza', 'condiciones', 'estado', 'funcionamiento']):
        return 'Condiciones del vehículo'
    elif 'punto de recogida' in texto or 'punto de entrega' in texto:
        return 'Lugar de entrega o recogida'
    elif 'gasolina' in texto or 'tanqueado' in texto:
        return 'Nivel de gasolina'
    elif 'papeles' in texto:
        return 'Papeles al día'
    elif 'seguro' in texto:
        return 'Seguro del vehículo'
    elif 'mantenimiento' in texto:
        return 'Mantenimiento'
    elif 'historial' in texto or 'kilometraje' in texto:
        return 'Historial del vehículo'
    elif 'confianza' in texto:
        return 'Confianza'
    elif 'agilidad' in texto:
        return 'Agilidad en la renta'
    elif 'contrato' in texto or 'condiciones claras' in texto:
        return 'Condiciones del contrato'
    elif 'vehículo nuevo' in texto or 'preferiblemente nuevo' in texto:
        return 'Vehículo nuevo'
    elif 'movimiento' in texto:
        return 'Rapidez en la plataforma'
    else:
        return 'Otro'

# Paso 2: Aplicar limpieza
df["P48N_Limpio"] = df["P48N-Otros_Puntos_Confianza_para_Rentar[P41]"].apply(clasificar_factores_confianza)

# Paso 3: Contar categorías (excluyendo 'Sin respuesta')
conteo = df["P48N_Limpio"].value_counts().drop("Sin respuesta", errors='ignore')

# Paso 4: Graficar
plt.figure(figsize=(10, 6))
conteo.plot(kind='bar', color='skyblue')
plt.title("Factores importantes al rentar un vehículo")
plt.xlabel("Categorías")
plt.ylabel("Cantidad de respuestas")
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
from collections import Counter

# Cargar archivo
df = pd.read_excel('../../docs/FLEX DRIVE - Permítenos Conocerte.xlsx', sheet_name='Sheet1')

# --- LIMPIEZA AVANZADA ---
def limpiar_comunicacion_ajustada(respuesta):
    if pd.isna(respuesta):
        return []

    respuesta = respuesta.lower()
    opciones = [op.strip().rstrip('.') for op in respuesta.split(';') if op.strip()]

    resultado = []

    for op in opciones:
        if "whatsapp" in op and "asesor" in op:
            resultado.append("Chat por WhatsApp")
        elif "chatbot" in op:
            resultado.append("Chatbot")
        elif "plataforma" in op and "chat" in op:
            resultado.append("Chat en plataforma")
        elif "correo" in op:
            resultado.append("Correo electrónico")
        elif "llamada" in op:
            resultado.append("Llamada telefónica")
        elif "no me interesa" in op or "ninguna" in op:
            return ["No le interesa comunicarse"]
        else:
            resultado.append("Otro")

    return list(set(resultado))

# Aplicar limpieza
df['P51_Comunicacion_Limpia'] = df['P51-Preferencias_para_Comunicarse_con_FlexDrive'].apply(limpiar_comunicacion_ajustada)

# Conteo final
conteo = Counter([item for sublist in df['P51_Comunicacion_Limpia'] for item in sublist])

# Mostrar conteo
print("📊 Conteo final de preferencias de comunicación:")
for canal, cantidad in conteo.items():
    print(f"{canal}: {cantidad}")

# Gráfica
plt.figure(figsize=(10,6))
plt.bar(conteo.keys(), conteo.values(), color='mediumseagreen')
plt.title("Preferencias de Comunicación con Flex Drive")
plt.xlabel("Canal de Comunicación")
plt.ylabel("Número de Respuestas")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
