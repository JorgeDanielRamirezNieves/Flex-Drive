```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df_FlexDrive = pd.read_csv('../../docs/df_FlexDrive_limpio.csv', sep=';')
```

```python
df_FlexDrive['P2-Edad'].value_counts().plot(kind='bar', figsize=(10, 6), color='skyblue')
plt.title('[Título de la Gráfica]]')
plt.xlabel('[Título del Eje X]]')
plt.ylabel('[Título del Eje Y]]')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Probando Seaborn para graficar, seccionar respuestas múltiples para contestar P7, ignorando NaNs
df_FlexDrive['P7-Actividad_Actual_Multi'].str.split(';', expand=True).stack().reset_index(drop=True).explode().value_counts().plot(kind='bar', figsize=(10, 6), color='lightgreen')

#Colocar número sobre barras
for index, value in enumerate(df_FlexDrive['P7-Actividad_Actual_Multi'].str.split(';', expand=True).stack().reset_index(drop=True).explode().value_counts()):
    plt.text(index, value, str(value), ha='center', va='bottom')
    
# Arreglar marco de gráfica para que no se corte el título y etiquetas


plt.title('[Título de la Gráfica]]')
plt.xlabel('[Título del Eje X]]')
plt.ylabel('[Título del Eje Y]]')
plt.xticks(rotation=90)
plt.tight_layout()

plt.gca().axes.get_yaxis().set_visible(False)

# Asegurar que el eje y tiene el rango correcto
plt.ylim(0, 250)

plt.show()
```
