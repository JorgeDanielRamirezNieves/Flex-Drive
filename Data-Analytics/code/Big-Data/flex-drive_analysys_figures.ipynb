```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df_FlexDrive = pd.read_csv('../../docs/poblaciones/df_FlexDrive_limpio.csv', sep=';')
```

```python
df_FlexDrive['P2-Edad'].value_counts().plot(kind='bar', figsize=(10, 6), color='skyblue')
plt.title('[Título de la Gráfica]]')
plt.xlabel('[Título del Eje X]]')
plt.ylabel('[Título del Eje Y]]')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Probando Seaborn para graficar, seccionar respuestas múltiples para contestar P7, ignorando NaNs
df_FlexDrive['P7-Actividad_Actual_Multi'].str.split(';', expand=True).stack().reset_index(drop=True).explode().value_counts().plot(kind='bar', figsize=(10, 6), color='lightgreen')

#Colocar número sobre barras
for index, value in enumerate(df_FlexDrive['P7-Actividad_Actual_Multi'].str.split(';', expand=True).stack().reset_index(drop=True).explode().value_counts()):
    plt.text(index, value, str(value), ha='center', va='bottom')
    
# Arreglar marco de gráfica para que no se corte el título y etiquetas


plt.title('[Título de la Gráfica]]')
plt.xlabel('[Título del Eje X]]')
plt.ylabel('[Título del Eje Y]]')
plt.xticks(rotation=90)
plt.tight_layout()

plt.gca().axes.get_yaxis().set_visible(False)

# Asegurar que el eje y tiene el rango correcto
plt.ylim(0, 250)

plt.show()
```

```python
import matplotlib.pyplot as plt

# Conteo individual de respuestas (ya limpias)
servicios_usados = df_FlexDrive["P10-Servicios_Públicos_Usados_Multi"].str.split(";").explode().value_counts()

# Crear la gráfica de barras
plt.figure(figsize=(10, 6))
servicios_usados.plot(kind="bar", color="#4C72B0", edgecolor="black")
plt.title("Servicios Públicos de Transporte Más Usados", fontsize=14)
plt.xlabel("Servicio", fontsize=12)
plt.ylabel("Número de Personas", fontsize=12)
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.grid(axis='y', linestyle='--', alpha=0.7)

# Mostrar la gráfica
plt.show()
```

```python
import matplotlib.pyplot as plt

# Conteo de menciones individuales (ya mapeadas)
conteo_entidades = df_FlexDrive["P29N-Entidades_Donde_Rentó[P25]"].str.split(";").explode().value_counts()

# Filtrar valores vacíos, si es necesario
conteo_entidades = conteo_entidades[conteo_entidades.index != ""]

# Gráfica de barras horizontal
plt.figure(figsize=(10, 6))
conteo_entidades.sort_values().plot(kind='barh', color='#69b3a2')
plt.title("Entidades donde los participantes han rentado vehículos previamente")
plt.xlabel("Cantidad de menciones")
plt.ylabel("Entidades o plataformas")
plt.tight_layout()
plt.grid(axis='x', linestyle='--', alpha=0.6)
plt.show()
```
