## Funcionalidad para Limpieza de Datos Múltiples

```python
def limpieza_respuestas_multi(serie):
    return (
        serie
        .fillna('')  # reemplazar NaN
        .astype(str)
        .str.replace(r'\s*;\s*', ';', regex=True)  # Quitar espacios alrededor de ;
        .str.replace(r'\.;', ';', regex=True)      # Quitar punto antes del ;
        .str.replace(r';$', '', regex=True)        # Quitar ; final
        .str.replace(r'\.$', '', regex=True)       # Quitar punto final
        .str.strip()
        .str.split(';')                            # Separar por ;
    )


def limpieza_preguntas_multi(df, columna):
    serie_limpia = limpieza_respuestas_multi(df[columna])
    
    ocupaciones_exploded = serie_limpia.explode().str.strip()
    conteo_ocupaciones = ocupaciones_exploded.value_counts()
    
    print(f"{conteo_ocupaciones}\n\nTotal de respuestas únicas: {len(conteo_ocupaciones)}")
    return serie_limpia


def mapeo_respuestas(df, columna, mapeo):
    
    def reemplazar_respuestas(lista_original):
        resultado = []
        for item in lista_original:
            item = item.strip()
            if item in mapeo:
                resultado.extend(mapeo[item])
            else:
                resultado.append(item)
        return list(dict.fromkeys([i.strip() for i in resultado if i]))

    # Limpiar y dividir
    serie_limpia = limpieza_respuestas_multi(df[columna])
    
    # Mapear
    serie_mapeada = serie_limpia.apply(reemplazar_respuestas)
    
    # Unir de nuevo como string separado por ";"
    df[columna] = serie_mapeada.apply(lambda x: ';'.join(x))

    return df
```

## Limpieza de P4 - Ciudad de Origen

```python
# Mostrando las ciudades únicas en la columna 'P4-Ciudad' en orden alfabético 
df_FlexDrive['P4-Ciudad'] = df_FlexDrive['P4-Ciudad'].str.strip()  # Eliminar espacios en blanco al inicio y al final
df_FlexDrive['P4-Ciudad'] = df_FlexDrive['P4-Ciudad'].str.title()  # Capitalizar la primera letra de cada palabra
```

```python
# Lista de ciudades válidas, puedes comenzar con las más comunes en tu región
ciudades_validas = ['Bogotá', 'Medellín', 'Cali', 'Barranquilla', 'Bucaramanga', 'Manizales', 'Tunja', 'Villavicencio']

# Crear diccionario de corrección automática
diccionario_ciudades = {}
for ciudad in df_FlexDrive['P4-Ciudad'].unique():
    if pd.isnull(ciudad):
        diccionario_ciudades[ciudad] = ciudad  # Mantener NaN
        continue
    mejor_match, score = process.extractOne(ciudad, ciudades_validas)
    if score > 85:  # puedes ajustar el umbral según tu confianza
        diccionario_ciudades[ciudad] = mejor_match
    else:
        diccionario_ciudades[ciudad] = ciudad  # queda igual si no hay buen match

# Aplicar correcciones
pd.Series(diccionario_ciudades).to_csv("diccionario_ciudades.csv")
```

```python
# Cargar el diccionario de ciudades desde el CSV
diccionario_ciudades = pd.read_csv('diccionario_oficial_ciudades.csv', index_col=0).squeeze().to_dict()

#Mostrar el diccionario de ciudades
print(diccionario_ciudades)

# Aplicar el diccionario de corrección a la columna 'P4-Ciudad'
df_FlexDrive['P4-Ciudad'] = df_FlexDrive['P4-Ciudad'].map(diccionario_ciudades)
# Mostrar las primeras filas del DataFrame con la columna 'P4-Ciudad' corregida
df_FlexDrive["P4-Ciudad"].value_counts().sort_index(ascending=True)
```
