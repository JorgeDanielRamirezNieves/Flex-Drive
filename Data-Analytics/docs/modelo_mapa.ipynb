```python
import folium
from geopy.geocoders import Nominatim
import time

# Datos
ciudades = {
    "Tunja (Boyacá)": 170,
    "Duitama (Boyacá)": 50,
    "Bogotá D.C.": 46,
    "Sogamoso (Boyacá)": 24,
    "Yopal (Casanare)": 21,
    "Ventaquemada (Boyacá)": 5,
    "Paz De Río (Boyacá)": 5,
    "Bucaramanga (Santander)": 3,
    "Medellín (Antioquia)": 3,
    "Cómbita (Boyacá)": 3,
    "Rusia": 3,
    "Barbosa (Santander)": 3,
    "Santa Rosa De Viterbo (Boyacá)": 3,
    "Cúcuta (Norte De Santander)": 2,
    "Tibasosa (Boyacá)": 2,
    "Tasco (Boyacá)": 2,
    "Floridablanca (Santander)": 2,
    "Socha (Boyacá)": 2,
    "Samacá (Boyacá)": 2,
    "Barranquilla (Atlántico)": 2,
    "Boyacá (Boyacá)": 2,
    "Barrancabermeja (Santander)": 2,
    "Paipa (Boyacá)": 2,
    "Armenia (Quindío)": 1,
    "Funza (Cundinamarca)": 1,
    "Puerto Boyacá (Boyacá)": 1,
    "La Capilla (Boyacá)": 1,
    "Guateque (Boyacá)": 1,
    "Nobsa (Boyacá)": 1,
    "San Juan Bosco (Boyacá)": 1,
    "Chía (Cundinamarca)": 1,
    "Cajicá (Cundinamarca)": 1,
    "Facatativá (Cundinamarca)": 1,
    "Villanueva (Casanare)": 1,
    "Gold Coast (Australia)": 1,
    "Kazan (Rusia)": 1,
    "Guanajuato (México)": 1,
    "Tenjo (Cundinamarca)": 1,
    "Catacamas (Honduras)": 1,
    "Ciudad de Guatemala (Guatemala)": 1,
    "El Carmen (Norte De Santander)": 1,
    "Riobamba (Ecuador)": 1,
    "Fusagasuga (Cundinamarca)": 1,
    "Villavicencio (Meta)": 1,
    "El Tarra (Norte De Santander)": 1,
}

# Crear el mapa centrado en Colombia
m = folium.Map(location=[5.0, -73.0], zoom_start=6)

# Geolocalizador
geolocator = Nominatim(user_agent="geoapi")

# Añadir cada ciudad al mapa
for ciudad, cantidad in ciudades.items():
    try:
        location = geolocator.geocode(ciudad, timeout=10)
        if location:
            folium.CircleMarker(
                location=[location.latitude, location.longitude],
                radius=3 + cantidad**0.5,  # radio proporcional
                popup=f"{ciudad}: {cantidad}",
                color="blue",
                fill=True,
                fill_opacity=0.6
            ).add_to(m)
            time.sleep(1)  # evitar bloquear por muchas peticiones
    except Exception as e:
        print(f"No se pudo geolocalizar: {ciudad} → {e}")

# Guardar el mapa en HTML
m.save("mapa_usuarios_flexdrive.html")
print("✅ Mapa generado como 'mapa_usuarios_flexdrive.html'")
```
